# Interdisciplinarity Metric Validation Notebook

This notebook is used to perform validation of interdisciplinarity metrics used in the Master's Thesis

The workflow is illustrated in the following blocks of code

```python
import numpy as np
import pandas as pd
import random
import math
import matplotlib.pyplot as plt
from collections import Counter
from scipy.stats import spearmanr, pearsonr
```

## Metric definitions

```python
def interdisciplinarity_from_counts(counts: dict) -> float:
    """Our definition of II and OI. Compute interdisciplinarity metric from a dict {category: count}."""
    vals = [v for v in counts.values() if v > 0]
    if not vals:
        return 0.0
    max_val = max(vals)
    sum_val = sum(vals)
    return 1 - max_val / sum_val

def shannon_entropy(counts: dict) -> float:
    """Normalized Shannon entropy."""
    vals = [v for v in counts.values() if v > 0]
    if not vals:
        return 0.0
    s = sum(vals)
    probs = [v/s for v in vals]
    H = -sum(p * math.log2(p) for p in probs)
    return H / math.log2(len(probs)) if len(probs) > 1 else 0.0

def simpson_index(counts: dict) -> float:
    """Simpson diversity index (1 - sum p^2)."""
    vals = [v for v in counts.values() if v > 0]
    if not vals:
        return 0.0
    s = sum(vals)
    probs = [v/s for v in vals]
    return 1 - sum(p**2 for p in probs)
```

## Bootstrap for Confidence Intervals

```python
def bootstrap_interdisciplinarity(works, B=500):
    """
    works: list of subfield labels for each work.
    Returns bootstrap median, 95% CI.
    """
    n = len(works)
    if n == 0:
        return (0, 0, 0)
    res = []
    for _ in range(B):
        sample = [random.choice(works) for _ in range(n)]
        counts = Counter(sample)
        res.append(interdisciplinarity_from_counts(counts))
    res = sorted(res)
    low = np.percentile(res, 2.5)
    high = np.percentile(res, 97.5)
    med = np.median(res)
    return (low, med, high)
```

## Null distribution

```python
def permutation_null(works, B=500):
    """
    Shuffle subfield labels to compute null distribution of metric.
    Returns array of null metrics.
    """
    n = len(works)
    if n == 0:
        return []
    res = []
    for _ in range(B):
        shuffled = random.sample(works, len(works))
        counts = Counter(shuffled)
        res.append(interdisciplinarity_from_counts(counts))
    return res
```

## Dataset upload (example)

```python
import json
from collections import Counter

# -------------------------------
# 4. Load JSON dataset (new structure)
# -------------------------------

with open("../authors_data/authors.json", "r", encoding="utf-8") as f:
    data = json.load(f)

currentYear = 2023
mode = "entire"     # "year" or "entire"
view = "subfields"  # "subfields" or "fields"

authors = {}  # dict: {authorId: list of works}

for author in data:
    works = []

    if view == "subfields":
        yearly = author.get("Yearly_Subfields", {})

        if mode == "year":
            counts = yearly.get(str(currentYear), {})
        else:
            # cumulative up to currentYear
            counts = {}
            for y, pairs in yearly.items():
                y = int(y)
                if y > currentYear: continue
                for subfield, val in pairs.items():
                    counts[subfield] = counts.get(subfield, 0) + int(val)

        # expand to repeated works
        for subfield, count in counts.items():
            works.extend([subfield] * count)

    elif view == "fields":
        yearly = author.get("Yearly_Fields", {})

        if mode == "year":
            counts = yearly.get(str(currentYear), {})
        else:
            # cumulative up to currentYear
            counts = {}
            for y, pairs in yearly.items():
                y = int(y)
                if y > currentYear: continue
                for pair, val in pairs.items():
                    counts[pair] = counts.get(pair, 0) + int(val)

        # expand: here you can choose either subfield or field
        for pair, count in counts.items():
            subfield, field = pair.split("---")
            works.extend([field] * count)       # focus on FIELD
            # or: works.extend([subfield] * count)  # focus on SUBFIELD

    authors[author["ORCID"]] = works

print("Prepared authors dict (works count):")
print({k: len(v) for k,v in authors.items()})
```

## Validation Workflow

```python
records = []
for name, works in authors.items():
    counts = Counter(works)
    I = interdisciplinarity_from_counts(counts)
    H = shannon_entropy(counts)
    S = simpson_index(counts)
    low, med, high = bootstrap_interdisciplinarity(works, B=500)
    records.append({
    "author": name,
    "n_works": len(works),
    "I": I,
    "H": H,
    "Simpson": S,
    "I_CI_low": low,
    "I_CI_med": med,
    "I_CI_high": high
    })

df = pd.DataFrame(records)
print("Validation Table")
print(df)
```

## Correlation Analysis

```python
print("\nCorrelations:")
print("I vs H (Spearman):", spearmanr(df["I"], df["H"]))
print("I vs Simpson (Spearman):", spearmanr(df["I"], df["Simpson"]))
print("I vs n_works (Spearman):", spearmanr(df["I"], df["n_works"]))
```

## Null test (for one example)

```python
example = authors["0000-0002-0472-0318"]
obs = interdisciplinarity_from_counts(Counter(example))
nulls = permutation_null(example, B=500)

plt.hist(nulls, bins=20, alpha=0.7, label="Null")
plt.axvline(obs, color="red", linestyle="--", label=f"Observed={obs:.2f}")
plt.legend()
plt.title("Permutation Null Test (Alice)")
plt.show()
```

## Metric vs Work Count (visualization)

```python
plt.scatter(df["n_works"], df["I"], label="I")
plt.scatter(df["n_works"], df["H"], label="Shannon")
plt.scatter(df["n_works"], df["Simpson"], label="Simpson")
plt.xlabel("Number of Works")
plt.ylabel("Metric Value")
plt.legend()
plt.title("Metrics vs Productivity")
plt.show()
```
